```python
rede = {
    "Iris": ["Gaby", "Ana"],
    "Ana": ["Gaby", "Iris"],
    "Laura": ["Gaby", "Miguel", "Juan"],
    "Alan": ["Juan", "Gaby", "Davi"],
    "Romeo": ["Rafa", "Juan"],
    "Rafa": ["Romeo", "Gaby", "Juan", "Maria"],
    "Maria": ["Rafa", "Juan"],
    "Santos": ["Lucas", "Miguel"],
    "Lucas": ["Miguel", "Santos"],
    "Iago": ["Miguel", "Kaique", "Luana"],
    "Luana": ["Iago", "Miguel", "Kaique"],
    "Kaique": ["Miguel", "Iago", "Luana"],
    "Pedro": ["Vitor", "Gaby", "Juan"],
    "Gaby": ["Iris", "Ana", "Laura", "Alan", "Rafa", "Juan", "Miguel", "Vitor", "Pedro"],
    "Juan": ["Romeo", "Rafa", "Maria", "Davi", "Miguel", "Vitor", "Pedro", "Gaby", "Laura", "Alan"],
    "Davi": ["Alan", "Juan", "Vitor", "Miguel"],
    "Miguel": ["Santos", "Lucas", "Iago", "Luana", "Kaique", "Vitor", "Gaby", "Laura", "Juan", "Davi"],
    "Vitor": ["Pedro", "Gaby", "Juan", "Davi", "Miguel"]
}

tamanho = len(rede)

for usuario, amigos in rede.items():
    print(usuario, "tem", len(amigos), "conexoes")

soma_graus = sum(len(amigos) for amigos in rede.values())
arestas = soma_graus // 2
densidade = (2 * arestas) / (tamanho * (tamanho - 1))

print("Total de usuarios:", tamanho)
print("Total de amizades:", arestas)
print(f"Densidade: {densidade:.2%}")
```
